# Part B — Dictionary Practice (25 points)


### 3. Create a dictionary called `class_points` using the following mapping. **(8 points)**

- First-Year → 10
- Sophomore → 20
- Junior → 30
- Senior → 40


````python
class_points = {
    'First-Year': 10,
    'Sophomore': 20,
    'Junior': 30,
    'Senior': 40
}
````

### 5. Use `.map()` to create a new column called `ClassPoints` from `students["ClassYear"]`.  
Then display `ClassYear` and `ClassPoints` for the first 10 students. **(7 points)**


````python
# Create the new column using .map() with the dictionary
students['ClassPoints'] = students['ClassYear'].map(class_points)

# Display ClassYear and ClassPoints for the first 10 students
students[['ClassYear', 'ClassPoints']].head(10)
````

# Part D — Function with Conditional Logic (30 points)


### 9. Create a function called `exam_category(score)`. **(12 points)**

Use these rules:

- score >= 90 → `"Excellent"`
- score >= 80 → `"Good"`
- score >= 70 → `"Fair"`
- score < 70 → `"Needs Improvement"`


````python
def exam_category(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 70:
        return "Fair"
    else:
        return "Needs Improvement"

````

### 10. Test `exam_category()` using the values `94`, `83`, `76`, and `62`. **(6 points)**

Before running the code, predict all four outputs.


````python

print(exam_category(94))
print(exam_category(83))
print(exam_category(76))
print(exam_category(62))
````

### 11. Use `.apply()` to create a new column called `ExamCategory` from `FinalExamScore`.  
Display `StudentID`, `FinalExamScore`, and `ExamCategory` for the first 10 students. **(7 points)**


````python

students['ExamCategory'] = students['FinalExamScore'].apply(exam_category)

students[['StudentID', 'FinalExamScore', 'ExamCategory']].head(10)
````

### 12. Create a second function called `study_category(hours)`. **(5 points)**

Use these rules:

- hours >= 15 → `"High"`
- hours >= 8 → `"Moderate"`
- hours < 8 → `"Low"`

Apply it to `StudyHoursPerWeek` to create a new column called `StudyCategory`.


````python
def study_category(hours):
    if hours >= 15:
        return "High"
    elif hours >= 8:
        return "Moderate"
    else:
        return "Low"

# Apply the function to create the new column
students['StudyCategory'] = students['StudyHoursPerWeek'].apply(study_category)

````
