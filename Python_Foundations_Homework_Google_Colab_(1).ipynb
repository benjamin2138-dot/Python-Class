### 18. Final Challenge

Your instructor wants to identify students who may need additional academic support.

Find students who:

- have `AttendancePct < 80`
- **AND**
- have `FinalExamScore < 75`

Then sort these students from **lowest to highest `FinalExamScore`**.

Display only:

- `StudentID`
- `Major`
- `AttendancePct`
- `FinalExamScore`

Then answer:

**How many students meet these criteria?**

````python
students_needing_support = students[
    (students["AttendancePct"] < 80) &
    (students["FinalExamScore"] < 75)
].sort_values("FinalExamScore", ascending=True)[
    ["StudentID", "Major", "AttendancePct", "FinalExamScore"]
]

students_needing_support
````

````python
len(students_needing_support)
````
